## Nombres de Harshad

```python
def est_harshad(n):
    n_str=str(n)
    s=0
    for i in n_str: 
        s+=int(i)
    if s==0 or n%s!=0:
        return False 
    return True
```

```python
est_harshad(21)
```

```python
est_harshad(19)
```

```python
est_harshad(72)
```

```python
est_harshad(0)
```

```python
def moitie(n):
    return n//2

def estdivisible(n,d):
    if n%d==0:
        return True
    else:
        return False

def sommediviseurs(n):
    liste=[]
    n_demi=moitie(n)
    for i in range(1,moitie(n)+1):
        if estdivisible(n,i):
            liste.append(i)
    somme=0
    for i in liste:
        somme+=i
    return somme

def sont_amis(n,m):
    somme_n=sommediviseurs(n)
    somme_m=sommediviseurs(m)
    if somme_n==m and somme_m==n:
        return True
    else:
        return False
    
```

```python
sont_amis(220,284)
```
