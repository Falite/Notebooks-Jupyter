```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
xp=np.linspace(1,4,1000)

def f(x):
    return x**3-2*x-5

plt.plot(xp,f(xp))
plt.show()
```

```python
def dichotomie(f,a,b,epsilon):
    it=0
    while b-a>epsilon:
        m=(a+b)/2
        it+=1
        if f(a)*f(m)<0:
            b=m
        else:
            a=m
    return [(a+b)/2, it]
```

```python
dichotomie(f,2,3,0.000001)
```
