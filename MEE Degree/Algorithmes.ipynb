## Nombres de Harshad

```python
def est_harshad(n):
    n_str=str(n)
    s=0
    for i in n_str: 
        s+=int(i)
    if s==0 or n%s!=0:
        return False 
    return True
```

```python
def moitie(n):
    return n//2

def estdivisible(n,d):
    if n%d==0:
        return True
    else:
        return False

def sommediviseurs(n):
    liste=[]
    n_demi=moitie(n)
    for i in range(1,moitie(n)+1):
        if estdivisible(n,i):
            liste.append(i)
    somme=0
    for i in liste:
        somme+=i
    return somme

def sont_amis(n,m):
    somme_n=sommediviseurs(n)
    somme_m=sommediviseurs(m)
    if somme_n==m and somme_m==n:
        return True
    else:
        return False
    
```

```python
def estabondant(n):
    somme=sommediviseurs(n)
    if somme>n : 
        return True
    else :
        return False

def listeabondant(n):
    liste=[]
    n_demi=moitie(n)
    if estabondant(n):
        for i in range(1,moitie(n)+1):
            if estdivisible(n,i):
                liste.append(i)
    return liste
```

### Faire une liste de nombre abondants d'un minimum à un maximum

```python
def listeabondants(n_min,n_max):
    if n_min > n_max:
        n_min2=n_max
        n_max2=n_min
        n_min=n_min2
        n_max=n_max2
    liste_abondants=[]
    for i in range(n_min,n_max):
        if estabondant(i):
            liste_abondants.append(i)
    return liste_abondants
```

```python
listeabondants(100,200)
```

```python
listeabondants(200,100)
```
