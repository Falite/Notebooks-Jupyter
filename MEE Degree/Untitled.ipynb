```python
def nombre(n):
    return n**2-n+11
```

```python
def moitie(n):
    return n//2

def estdivisible(n,d):
    if n%d==0:
        return True
    else:
        return False

def estpremier(n):
    for i in range(1,moitie(n)):
        if i!=1 and estdivisible(n,i):
            return False
    return True
```

```python
for i in range(2,150):
    m=i**2-i+11
    if estpremier(m):
        print("pour n=",i," on a n^2-n+11=",m," et ",m," est premier.")
```
