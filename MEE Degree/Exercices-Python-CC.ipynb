```python
xp=np.linspace(1,4,1000)

def f(x):
    return x**3-2*x-5

plt.plot(xp,f(xp))
plt.show()
```

```python
def dichotomie(f,a,b,epsilon):
    it=0
    while b-a>epsilon:
        m=(a+b)/2
        it+=1
        if f(a)*f(m)<0:
            b=m
        else:
            a=m
    return [(a+b)/2, it]
```

```python
dichotomie(f,2,3,0.000001)
```

## Tortue 

```python
import turtle as t
```

```python
t.forward(60)
```

```python
def koch(longueur,n):
    if n==0:
        t.forward(longueur)
    else:
        koch(longueur/3, n-1)
        t.left(60)
        koch(longueur/3, n-1)
        t.right(120)
        koch(longueur/3, n-1)
        t.left(60)
        koch(longueur/3, n-1)
```

```python
koch(900, 2)
```

## Jeu de Loterie
#### Mardi 29 Septembre 2026


```python
def jeuloterie(mise):
    secteur1=random.randint(1,8)
    secteur2=0
    gain=0
    if secteur1==1:
        gain=2*mise
    else:
        secteur2=random.randint(1,8)
        if secteur2==1 or secteur2==2 or secteur2==3 or secteur2==4:
            gain=mise+6
        else:
            gain=0
    return secteur1, secteur2, gain
```

```python
for i in range(1,100):
    mise=random.randint(20,400)
    resultatjeu=jeuloterie(mise)
    if resultatjeu[0]==1:
        print("La mise était de ",mise," et la roue est tombée sur le secteur 1, jackpot ! Le gain est de ", resultatjeu[2])
    else:
        print("La mise était de ",mise," et la roue est tombée une première fois sur le secteur ",resultatjeu[0],". On relance la roue et elle est tombée sur le secteur ",resultatjeu[1],". Le gain  est de ", resultatjeu[2],"€")
```

```python
for i in range(1,100):
    resultatjeu=jeuloterie(30)
    if resultatjeu[0]==1:
        print("La mise était de 30€ et la roue est tombée sur le secteur 1, jackpot ! Le gain est de ", resultatjeu[2])
    else:
        print("La mise était de 30€ et la roue est tombée une première fois sur le secteur ",resultatjeu[0],". On relance la roue et elle est tombée sur le secteur ",resultatjeu[1],". Le gain  est de ", resultatjeu[2],"€")
```

```python
xp=np.arange(1,100,1)
yp=[]
for i in range(1,100):
    mise=30
    yp.append(jeuloterie(mise)[2]-30)

plt.plot(xp,yp,"+")
plt.show()
```

```python
echantillon=1000
mise=9
parties=50

E=[]
for i in range(1,echantillon):
    m=0
    for j in range(1,parties):
        m+=jeuloterie(mise)[2]
    E.append(m/parties)

plt.plot(E,'+',color="yellow")
plt.axhline(y=mise, color='red', linestyle='-', linewidth=1)
plt.axhline(y=mise-2.5, color='blue', linestyle='--', linewidth=1)
plt.axhline(y=mise+2.5, color='blue', linestyle='--', linewidth=1)
plt.show()
```
