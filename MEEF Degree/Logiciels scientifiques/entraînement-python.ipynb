```python
import numpy as np
from matplotlib import pyplot as plt
```

### Représentation 1

```python
def f(x):
    return x*np.sin(1/x)

xmin,xmax = 1,2
x = np.linspace(xmin,xmax,100)
plt.figure(figsize=(10,8))
plt.plot(x,f(x))
```

### Représentation graphique 2

```python
def f(k,x):
    return x**k*np.log(x+2)

xmin,xmax= -1.5,2
x = np.linspace(xmin,xmax,100)

for k in range(5):
    plt.plot(x,f(k,x))

plt.show()
```

### Génération d'une suite définie explicitement

```python
def u(n):
    return 3*n**2+n-1

vec_n=np.arange(0,19,1)
plt.plot(vec_n,u(vec_n),'+')
plt.show()
```

### Génération d'une suite définie par récurrence d'ordre 1

```python
def u_rec(n) :
    if 0 == n:
        return 2
    return 2*u_rec(n-1)-1

plt.plot(vec_n,[u_rec(n) for n in vec_n],'+')
plt.show()
```

### Génération d'une suite définie par récurrence d'ordre 2

```python
def Fibonacci(n):
    if n == 0 or n == 1:
        return 1
    return Fibonacci(n-1) + Fibonacci(n-2)

plt.plot(vec_n,[Fibonacci(n) for n in vec_n],'+')
plt.show()
```

### Simulation d'une expérience aléatoire

```python
def descubiques(n):
    liste_dix = []
    liste_neuf = []
    for i in range(n):
        x=np.random.randint(1,7)
        y=np.random.randint(1,7)
        s=x+y
        if s==10:
            liste_dix.append(1)
        elif s==9:
            liste_neuf.append(1)
    return len(liste_dix),len(liste_neuf)

r=descubiques(100)

print("On a obtenu ",r[0]," fois le nombre 10 et ",r[1]," fois le nombre 9.")
```
