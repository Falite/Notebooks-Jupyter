### Question 1 :

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sin(x)**2
def g(x):
    return np.abs(x)*x

xmin,xmax=-np.pi, np.pi
x=np.linspace(xmin,xmax,1000)

plt.plot(x,f(x))
plt.plot(x,g(x))
plt.show()
```

### Question 2

```python
def u(n):
    if n==0:
        return 1
    else:
        return u(n-1)+ 3/u(n-1)

vec_n=np.arange(1,10)
plt.plot(vec_n,[u(n) for n in vec_n],'+')
plt.show()
```
