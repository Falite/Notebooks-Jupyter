### Question 1

```python
import numpy as np
import matplotlib.pyplot as plt

xmin, xmax = -np.pi, np.pi
xp=np.linspace(xmin, xmax, 100)

def f(x):
    return np.sqrt(x*np.sin(x))
def g(x):
    return -np.sqrt(np.pi**2-np.pi*np.abs(x))

plt.plot(xp,f(xp),color='b',label='$f(x)$')
plt.plot(xp,g(xp),color='r',label='$g(x)$')
plt.show()
```

### Question 2

```python
#fonction pour simuler un seul lancer de dé
def lancer_dé ():
    de=[1,2,3,4,5,6]
    return np.random.choice(de, p=[0.2,0.05,0.1,0.25,0.05,0.35])

#simulation de 10 lancers de dé
L_res = []
for i in range(10):
    L_res.append(lancer_dé())

for k in range(len(L_res)):
    print('le ',k+1,'ème lancer donne',L_res[k])
```

### Question  3

```python
def u(n):
    if n==0:
        return 1
    else:
        return np.sqrt((u(n-1)+13)/u(n-1))

vec_n = np.arange(0,10,1)

plt.plot(vec_n,[u(n) for n in vec_n],'+',color='g')
plt.show()
```
