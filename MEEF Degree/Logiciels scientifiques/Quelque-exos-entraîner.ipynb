```python
def piecetruque():
    piece=["pile","face"]
    return np.random.choice(piece, p=[0.8,0.2])

r=piecetruque()
print(r)
```

##### Question 5

```python
def lancerspiecetruque(n):
    liste_pile=[]
    liste_face=[]
    for i in range(n):
        r=piecetruque()
        if r=="pile":
            liste_pile.append(i)
        else:
            liste_face.append(i)
    return len(liste_face), len(liste_pile)

R=lancerspiecetruque(10)
print("on a obtenu ", R[0], " fois face et ", R[1], " fois pile.")
```

- Il y a 6 clous donc si elle atteint 6, la bille n'a fait que tomber à droite donc elle est forcément dans la case n°6.
- Il y a 6 clous donc si elle atteint 3, la bille est exactement tombé 3 fois à droite et 3 autres fois à gauche donc quel que soit le chemin elle tombe dans la case n°3.
- Il y a 6 clous donc si elle reste à 0, la bille n'est pas du tout tombé à droite donc est à chaque fois tombée à gauche, donc elle est forcément dans la case n°0.

##### Question 1b

```python
from random import *

def simul_une_bille(h):
    droite=0
    for i in range(h):
        droite+=randint(0,1)
    return droite

simul_une_bille(6)
```

##### Question 1c

```python
def simul_N_billes(N,h):
    planche=[0 for i in range(h+1)]
    for i in range(N):
        a=simul_une_bille(h)
        planche[a]=planche[a]+1
    return(planche)

simul_N_billes(100,6)
```
