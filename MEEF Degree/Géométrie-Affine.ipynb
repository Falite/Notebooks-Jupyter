Soit $E$ un espace vectoriel euclidien de dimension 3.\
Déterminer la matrice de l'endomorphisme $f$ de $E$ de matrice relativement à une base orthonormée directe $(i,j,k)$ de $E$ (et préciser les éléments caractéristiques.\
$A=\frac 1{27} \begin{pmatrix} 2 & -26 & 7 \\ -23 & 2 & 14 \\ 14 & 7 & 22 \end{pmatrix}$ \
NB : Pour une anti rotation (rotation+réflexion) la trace est $-1+2 \cos \theta$ \
Car dans une base convenable, on aurait $\begin{pmatrix} -1 & 0 & 0 \\ 0 & \cos \theta & - \sin \theta \\ 0 & \sin \theta & \cos \theta \end{pmatrix}$

## Résolution

On écrit la matrice $A$

```python
A_0 = matrix([[2,-26,7],[-23,2,14],[14,7,22]])
```

```python
A = (1/27)*A_0
A
```

En faisant $A \times ^t A $ on trouve que c'est bien $I_3$ donc la matrice est bien orthogonale

```python
(A * transpose(A)) 
```

En calculant le déterminant on a $-1$ ça correspont à un anti déplacement.

```python
A.det()
```

On peut aussi rechercher ses valeurs propres : 

```python
vpropres=A.eigenvalues()
vpropres
```

On résout ensuite $(A-I)X=0_3$ \
En posant $(x,y,z) \in \mathbb R^3$

```python
A_solv = A - identity_matrix(3)
```

```python
A_solv = 27*A_solv
A_solv
```

```python
A_solv.det()
```

La matrice est inversible donc le système peut être résolu. 

```python
var('x y z')

X=vector([x, y, z])
```

```python
solution = A_solv.right_kernel()
print("Solution de (A-I)*X = 0 :", solution)
```

On a $AX=X \Leftrightarrow 27 A X = 27 X \Leftrightarrow 27AX-27X=0$
On trouve que l'équation $(A-I)X = 0$ a pour solution le singleton $\{ (0,0,0)\}$.\
C'est donc une anti rotation.

```python
A_solv.inverse()
```

```python
angle = arccos((A.trace()+1)/2)
angle
```

On calcule l'axe de rotation : 

```python
A_rot = 27*(A + identity_matrix(3))

rotation = A_rot.right_kernel()
print("Solution de (A+I)X=0 :", rotation)
```

l'axe de rotation est donc ce vecteur.
