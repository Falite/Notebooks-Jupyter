# Etude de la corrélatrion entre variables 

```python
list_corr=['Age','Academic Pressure','Work Pressure','CGPA','Study Satisfaction','Job Satisfaction','Work/Study Hours']

df_depressifs_corr=df_depressifs[list_corr]

matrice_depressifs_corr=df_depressifs_corr.corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(matrice_depressifs_corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.show()
```
