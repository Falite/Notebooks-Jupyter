## Nettoyage ou non des données

```python
df_vaccin.isnull().sum()
```

Il semble que des données manquent pour certains pays.

```python
len(df_vaccin['date'].unique())
```

Il y a à peu près 500 dates différentes pour chaque pays, voyons lesquels manquent de données.

```python
country = df_vaccin['country'].unique().tolist()

for c in country:
    df_temp=df_vaccin[df_vaccin['country']==c]
    plt.figure(figsize=(20,6))
    plt.plot(df_temp['total_vaccination'],bins=50)
    plt.title(f"vaccinations totale en {c}")
```
