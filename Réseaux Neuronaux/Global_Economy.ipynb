#### Explications des variables

1. ##### Indicateurs de Change et Monnaie

    - ```AMA exchange rate``` : Taux de change AMA
        Taux de change déterminé par l'accord de coopération monétaire entre les pays membres de l'AMA (Accord Monétaire Africain).
    - ```IMF based exchange rate``` : Taux de change basé sur le FMI
        Taux de change déterminé ou influencé par le Fonds Monétaire International (FMI).
    - ```Currency``` : Monnaie
        Unité monétaire utilisée dans un pays pour les transactions économiques.

2. ##### Indicateurs Démographiques

    - ```Population``` : Population
        Nombre total d'habitants dans un pays ou une région.

3. ##### Indicateurs de Revenu et Production

    - ```Per capita GNI``` : Revenu National Brut (RNB) par habitant
        Revenu total d'un pays divisé par sa population, indiquant le revenu moyen par personne.
    - ```Gross National Income (GNI) in USD``` : Revenu National Brut (RNB) en USD
        Revenu total d'un pays, incluant les revenus des résidents à l'étranger, converti en dollars américains.
    - ```Gross Domestic Product (GDP)``` : Produit Intérieur Brut (PIB)
        Valeur totale des biens et services produits dans un pays sur une période donnée.

4. ##### Composantes du PIB

    - ```Exports of goods and services``` : Exportations de biens et services
        Valeur des biens et services produits dans un pays et vendus à l'étranger.
    - ```Imports of goods and services``` : Importations de biens et services
        Valeur des biens et services achetés à l'étranger et consommés dans le pays.
    - ```Final consumption expenditure``` : Dépenses de consommation finale
        Dépenses totales des ménages, des administrations publiques et des institutions sans but lucratif au service des ménages pour l'achat de biens et services finaux.
    - ```Household consumption expenditure (including Non-profit institutions serving households)``` : Dépenses de consommation des ménages (y compris les institutions sans but lucratif au service des ménages)
        Dépenses des ménages pour l'achat de biens et services finaux.
    - ```General government final consumption expenditure``` : Dépenses de consommation finale des administrations publiques
        Dépenses des administrations publiques pour l'achat de biens et services finaux.
    - ```Gross capital formation``` : Formation brute de capital
        Investissements réalisés par les entreprises et les gouvernements pour augmenter la capacité de production future.
    - ```Gross fixed capital formation (including Acquisitions less disposals of valuables)``` : Formation brute de capital fixe (y compris les acquisitions moins les cessions de valeurs)
        Investissements dans des actifs fixes comme les bâtiments, les machines, etc., y compris les acquisitions moins les cessions de valeurs.
    - ```Changes in inventories``` : Variations des stocks
        Changements dans les niveaux de stocks de biens détenus par les entreprises.

5. ##### Secteurs Économiques

    - ```Agriculture, hunting, forestry, fishing (ISIC A-B)``` : Agriculture, chasse, sylviculture, pêche (ISIC A-B)
        Secteur économique regroupant les activités liées à l'agriculture, la chasse, la sylviculture et la pêche.
    - ```Mining, Manufacturing, Utilities (ISIC C-E)``` : Mines, manufactures, services publics (ISIC C-E)
        Secteur économique regroupant les activités liées à l'extraction minière, la fabrication et les services publics.
    - ```Manufacturing (ISIC D)``` : Fabrication (ISIC D)
        Secteur économique regroupant les activités de transformation des matières premières en produits finis.
    - ```Construction (ISIC F)``` : Construction (ISIC F)
        Secteur économique regroupant les activités de construction de bâtiments et d'infrastructures.
    - ```Wholesale, retail trade, restaurants and hotels (ISIC G-H)``` : Commerce de gros, commerce de détail, restaurants et hôtels (ISIC G-H)
        Secteur économique regroupant les activités de commerce de gros, de détail, de restauration et d'hébergement.
    - ```Transport, storage and communication (ISIC I)``` : Transport, entreposage et communication (ISIC I)
        Secteur économique regroupant les activités de transport, d'entreposage et de communication.
    - ```Other Activities (ISIC J-P)``` : Autres activités (ISIC J-P)
        Secteur économique regroupant diverses autres activités non classées ailleurs.
    - ```Total Value Added``` : Valeur ajoutée totale
        Valeur ajoutée créée par les différents secteurs économiques, calculée comme la différence entre la production totale et les consommations intermédiaires.

## Focus sur l'économie française

````python
pd.options.mode.chained_assignment = None
````

````python
df_eco_fr=df_eco[df_eco[' Country ']==' France ']
df_eco_fr.drop(' Country ',axis=1,inplace=True)
df_eco_fr.drop(' CountryID ',axis=1,inplace=True)
df_eco_fr.head()
````

### Indicateurs du Pib

````python
sns.pairplot(df_eco_fr[composantes_pib])
````

````python
Year=df_eco_fr[' Year '].tolist() 

#On commence par lister les crises pour les faire apparaître sur notre barplot
petrole_OPEP=Year.index(1973)

petrole_iran=Year.index(1979)

subprime_d=Year.index(2007)
subprime_f=Year.index(2008)

dette_européenne_d=Year.index(2009)
dette_européenne_f=Year.index(2012)

covid_d=Year.index(2020)
covid_f=Year.index(2021)

for elem in composantes_pib:
    plt.figure(figsize=(20,6))
    sns.barplot(x=df_eco_fr[' Year '], y=df_eco_fr[elem])
    #plt.axvline(x=subprime_d, color='red', linestyle='-', label='début de la Crise des subprimes')
    #plt.axvline(x=subprime_f, color='red', linestyle='-', label='fin de la Crise des subprimes')
    plt.axvline(x=petrole_OPEP, color='black', linestyle='-', label='Crise pétrolière embargo de OPEP')
    plt.axvline(x=petrole_iran, color='black', linestyle='-', label='Crise pétrolière de la révolution iranienne')
    plt.axvspan(subprime_d, subprime_f, color='red',alpha=0.3,  label='Crise des subprimes')
    plt.axvspan(dette_européenne_d, dette_européenne_f, color='blue', alpha=0.3, label='Crise de la dette européenne')
    plt.axvspan(covid_d, covid_f, color='yellow', alpha=0.3, label='Crise du Covid-19')
    
    titre=traduction.get(elem, elem)
    plt.title(f'{titre}')
    plt.xticks(rotation=45)
    plt.legend()

plt.show()
````
